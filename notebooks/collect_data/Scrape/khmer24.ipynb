```python
import requests
from bs4 import BeautifulSoup
import time
import random
import json
import re

# Configuration
HEADERS = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
    'Accept-Language': 'en-US,en;q=0.5'
}
DELAY = random.uniform(1.5, 3.0)  # Respectful delay between requests

def scrape_realestate_com_kh():
    """Scrape www.realestate.com.kh (buy section)"""
    print("Scraping realestate.com.kh...")
    url = "https://www.realestate.com.kh/buy/"
    results = []
    
    try:
        response = requests.get(url, headers=HEADERS, timeout=15)
        soup = BeautifulSoup(response.text, 'html.parser')
        
        for card in soup.select('div.listing-item'):
            title = card.select_one('h2.listing-title').get_text(strip=True) if card.select_one('h2.listing-title') else None
            price = card.select_one('div.price').get_text(strip=True) if card.select_one('div.price') else None
            location = card.select_one('div.listing-address').get_text(strip=True) if card.select_one('div.listing-address') else None
            features = [f.get_text(strip=True) for f in card.select('ul.listing-property-details li')]
            
            results.append({
                'title': title,
                'price': price,
                'location': location,
                'features': features,
                'url': card.select_one('a')['href'] if card.select_one('a') else None
            })
            
        time.sleep(DELAY)
    except Exception as e:
        print(f"Error scraping realestate.com.kh: {e}")
    
    return results

def scrape_kh24():
    """Scrape www.kh24.com.kh/property"""
    print("Scraping kh24.com.kh...")
    url = "https://www.kh24.com.kh/property"
    results = []
    
    try:
        response = requests.get(url, headers=HEADERS, timeout=15)
        soup = BeautifulSoup(response.text, 'html.parser')
        
        for item in soup.select('div.property-listing'):
            title = item.select_one('h3 > a').get_text(strip=True) if item.select_one('h3 > a') else None
            price = item.select_one('.property-price').get_text(strip=True) if item.select_one('.property-price') else None
            location = item.select_one('.property-address').get_text(strip=True) if item.select_one('.property-address') else None
            details = [d.get_text(strip=True) for d in item.select('.property-features li')]
            
            results.append({
                'title': title,
                'price': price,
                'location': location,
                'details': details,
                'url': item.select_one('a')['href'] if item.select_one('a') else None
            })
            
        time.sleep(DELAY)
    except Exception as e:
        print(f"Error scraping kh24.com.kh: {e}")
    
    return results

def scrape_propertyarea_asia():
    """Scrape www.propertyarea.asia/cambodia"""
    print("Scraping propertyarea.asia...")
    url = "https://www.propertyarea.asia/cambodia"
    results = []
    
    try:
        response = requests.get(url, headers=HEADERS, timeout=15)
        soup = BeautifulSoup(response.text, 'html.parser')
        
        for card in soup.select('div.property-item'):
            title = card.select_one('h2 > a').get_text(strip=True) if card.select_one('h2 > a') else None
            price = card.select_one('.price').get_text(strip=True) if card.select_one('.price') else None
            location = card.select_one('.location').get_text(strip=True) if card.select_one('.location') else None
            features = [f.get_text(strip=True) for f in card.select('.features-list li')]
            
            results.append({
                'title': title,
                'price': price,
                'location': location,
                'features': features,
                'url': card.select_one('a')['href'] if card.select_one('a') else None
            })
            
        time.sleep(DELAY)
    except Exception as e:
        print(f"Error scraping propertyarea.asia: {e}")
    
    return results

def scrape_bodiaproperties():
    """Scrape www.bodiaproperties.com"""
    print("Scraping bodiaproperties.com...")
    url = "https://www.bodiaproperties.com"
    results = []
    
    try:
        response = requests.get(url, headers=HEADERS, timeout=15)
        soup = BeautifulSoup(response.text, 'html.parser')
        
        for property in soup.select('div.property-item'):
            title = property.select_one('h4 > a').get_text(strip=True) if property.select_one('h4 > a') else None
            price = property.select_one('.price').get_text(strip=True) if property.select_one('.price') else None
            location = property.select_one('.location').get_text(strip=True) if property.select_one('.location') else None
            details = property.select_one('.property-details').get_text(strip=True) if property.select_one('.property-details') else None
            
            results.append({
                'title': title,
                'price': price,
                'location': location,
                'details': details,
                'url': property.select_one('a')['href'] if property.select_one('a') else None
            })
            
        time.sleep(DELAY)
    except Exception as e:
        print(f"Error scraping bodiaproperties.com: {e}")
    
    return results

if __name__ == "__main__":
    all_data = {
        "realestate_com_kh": scrape_realestate_com_kh(),
        "kh24": scrape_kh24(),
        "propertyarea_asia": scrape_propertyarea_asia(),
        "bodia_properties": scrape_bodiaproperties()
    }
    
    # Save results to JSON file
    with open('cambodia_real_estate.json', 'w', encoding='utf-8') as f:
        json.dump(all_data, f, ensure_ascii=False, indent=2)
    
    print("Scraping completed. Data saved to cambodia_real_estate.json")
```
